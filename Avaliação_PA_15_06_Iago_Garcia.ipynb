```python
# 2. Solicite ao usuário um número inteiro positivo. Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

num = int(input("Digite um número: "))
N = 0
for i in range(1, num + 1):
  N += i
print(f"A soma dos números até {num} é: {N}")
```

```python
# 4. Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

num = int(input("Digite um número inteiro positivo: "))

if num < 0:
    print("Fatorial não é definido para números negativos.")
elif num == 0:
    print("O fatorial de 0 é 1.")
else:
    fatorial = 1
    for i in range(1, num + 1):
        fatorial *= i
    print(f"O fatorial de {num} é {fatorial}.")
```

```python
# 5. Caixa eletrônico
vs = int(input("Digite o valor do saque: R$ "))

notas = [100, 50, 20, 10, 5, 2, 1]
contagem_notas = {}

vr = vs

print(f"Para sacar R$ {vs}, você receberá:")

for nota in notas:
    if vr >= nota:
        quantidade = vr // nota
        contagem_notas[nota] = quantidade
        vr %= nota
        print(f"{quantidade} nota(s) de R$ {nota}")

if vr > 0:
    print(f"\nSobrou um valor de R$ {vr} que não pode ser sacado com as notas disponíveis.")
```
