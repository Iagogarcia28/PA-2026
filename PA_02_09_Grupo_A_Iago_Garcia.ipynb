```python
# 4
notas = [7.5, 8.0, 6.5, 9.0, 8.5]
soma_notas = sum(notas)
media = soma_notas / len(notas)
maior_nota = max(notas)
menor_nota = min(notas)
print(f"Soma das notas: {soma_notas}")
print(f"Média do aluno: {media:.2f}")
print(f"Maior nota: {maior_nota}")
print(f"Menor nota: {menor_nota}")
if media >= 7.0:
    print("Situação: Aprovado!")
else:
    print("Situação: Reprovado.")
```

```python
# 5
numeros = list(range(1, 21))
pares = []
for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)

print("Lista de pares (loop for):", pares)
pares_comprehension = [numero for numero in numeros if numero % 2 == 0]
print("Lista de pares (list comprehension):", pares_comprehension)
```

```python
# 7
numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]


positivos = []
negativos = []
zeros = []


for num in numeros:
    if num > 0:
        positivos.append(num)
    elif num < 0:
        negativos.append(num)
    else:
        zeros.append(num)


print("Números positivos:", positivos)
print("Números negativos:", negativos)
print("Zeros (desafio):", zeros)
```

```python
# 8
numeros_originais = [42, 15, 88, 3, 27, 91, 50, 12, 74, 33]

print("Lista original:", numeros_originais)

numeros = numeros_originais.copy()


numeros.sort()
print("Ordem crescente:", numeros)

numeros.sort(reverse=True)
print("Ordem decrescente:", numeros)


maior = max(numeros_originais)
menor = min(numeros_originais)

print(f"Maior valor: {maior}")
print(f"Menor valor: {menor}")
```

```python
# 9
produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00]
]


print("--- LISTA DE PRODUTOS ---")
for produto in produtos:
    nome = produto[0]
    preco = produto[1]
    print(f"{nome}: R$ {preco:.2f}")


total = sum(produto[1] for produto in produtos)
print(f"\nValor total da compra: R$ {total:.2f}")


produto_mais_caro = max(produtos, key=lambda p: p[1])
produto_mais_barato = min(produtos, key=lambda p: p[1])

print(f"Produto mais caro: {produto_mais_caro[0]} (R$ {produto_mais_caro[1]:.2f})")
print(f"Produto mais barato: {produto_mais_barato[0]} (R$ {produto_mais_barato[1]:.2f})")


print("\n--- BUSCA DE PRODUTO ---")
busca = input("Digite o nome de um produto para buscar: ").strip().lower()

encontrado = False
for produto in produtos:

    if produto[0].lower() == busca:
        print(f"Produto encontrado! {produto[0]} custa R$ {produto[1]:.2f}.")
        encontrado = True
        break

if not encontrado:
    print("Produto não encontrado no cadastro.")
```

```python
# 10
qtd_alunos = int(input("Quantos alunos deseja cadastrar? "))

turma = []


for i in range(qtd_alunos):
    print(f"\n--- Cadastro do Aluno {i + 1} ---")
    nome = input("Nome do aluno: ").strip()
    nota = float(input(f"Nota de {nome}: "))


    turma.append([nome, nota])


print("\n" + "="*30)
print("RELATÓRIO DA TURMA")
print("="*30)


print("\nLista de Alunos e Notas:")
for aluno in turma:
    print(f"- {aluno[0]}: {aluno[1]:.1f}")


soma_notas = sum(aluno[1] for aluno in turma)
media_turma = soma_notas / len(turma)
print(f"\nMédia da turma: {media_turma:.2f}")


aluno_maior_nota = max(turma, key=lambda a: a[1])
aluno_menor_nota = min(turma, key=lambda a: a[1])

print(f"Maior nota: {aluno_maior_nota[0]} ({aluno_maior_nota[1]:.1f})")
print(f"Menor nota: {aluno_menor_nota[0]} ({aluno_menor_nota[1]:.1f})")


aprovados = [aluno[0] for aluno in turma if aluno[1] >= 7.0]
reprovados = [aluno[0] for aluno in turma if aluno[1] < 7.0]

print(f"\nAprovados (nota >= 7.0): {', '.join(aprovados) if aprovados else 'Nenhum'}")
print(f"Reprovados (nota < 7.0): {', '.join(reprovados) if reprovados else 'Nenhum'}")


print("\n" + "="*30)
print("CONSULTA DE ALUNO")
print("="*30)

busca = input("Digite o nome do aluno que deseja pesquisar: ").strip().lower()

encontrado = False
for aluno in turma:
    if aluno[0].lower() == busca:
        print(f"Aluno encontrado: {aluno[0]} | Nota: {aluno[1]:.1f}")
        encontrado = True
        break

if not encontrado:
    print("Aluno não encontrado no cadastro.")
```
