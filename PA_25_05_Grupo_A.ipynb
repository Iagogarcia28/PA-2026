```python
# Calculadora simples - escrever um programa que calcule as 4 operações matemáticas

# opção 1 - somar
# opção 2 - subtrair
# opção 3 - multiplicar
# opção 4 - dividir
# opção 5 - sair
opcao = ""
a = 0
b = 0
try:

  while opcao != 5:
    opcao = int(input("Digite a opção desejada ou 5 para sair: "))

    if opcao == 1:
      a = float(input("Digite o primeiro numero: "))
      b = float(input("Digite o segundo numero: "))
      print(f"Total da soma de {a} + {b} é igual a: {a + b}")
      print("")

    elif opcao == 2:
      a = float(input("Digite o primeiro numero: "))
      b = float(input("Digite o segundo numero: "))
      print(f"A subtração de {a} - {b} é igual a: {a - b}")
      print("")

    elif opcao == 3:
      a = float(input("Digite o primeiro numero: "))
      b = float(input("Digite o segundo numero: "))
      print(f"A multiplicação de {a} * {b} é igual a: {a * b}")
      print("")

    elif opcao == 4:
      a = float(input("Digite o primeiro numero: "))
      b = float(input("Digite o segundo numero: "))
      print(f"A divisão de {a} / {b} é igual a: {a / b}")
      print("")
except:
  print("Digite apenas números de 1 a 5")
```

```python
# escrever um programa que recebe 4 notas, calcula a média e imprimi se o aluno foi aprovado, recuperação ou reprovado
# nota maior ou igual a 7 - aprovado
# nota maior ou igual a 5 - recuperação
# nota menor que 5 - reprovado

try:
  nota1 = float(input("Digite a primeira nota: "))
  nota2 = float(input("Digite a segunda nota: "))
  nota3 = float(input("Digite a terceira nota: "))
  nota4 = float(input("Digite a quarta nota: "))

  media = (nota1 + nota2 + nota3 + nota4) / 4

  print(f"A média do aluno é: {media:.2f}")

  if media >= 7:
    print("Aluno APROVADO!")
  elif media >= 5:
    print("Aluno em RECUPERAÇÃO.")
  else:
    print("Aluno REPROVADO.")
except ValueError:
  print("Entrada inválida. Por favor, digite apenas números para as notas.")
```
