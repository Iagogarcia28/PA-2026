```python
# escrever um programa que o usuario digita um numero de 1 a 10 e imprime a tabuada
b = int(input("Digite um número de 1 a 10: "))
for i in range (1, 11):
  print(f"{i} X {b} = {i * b}")
```

```python
# 2 - Escrever um programa que o usuario digita um numero e o programa imprimi o numero digitado até o zero (contagem regressiva).
num = int(input("Digite um número: "))
for i in range(num, -1, -1):
  print(i)
```

```python
# 3 - Escrever um programa que o usuario digita um numero e o programa imprimi os numeros pares
num = int(input("Digite um número: "))
for i in range(1, num + 1):
  if i % 2 == 0:
    print(i)
```

```python
# 4 - Escrever um programa que o usuario digita um numero e o programa imprimi os numeros impares
num = int(input("Digite um número: "))
for i in range(1, num + 1):
  if i % 2 != 0:
    print(i)
```

```python
# 5 - Escrever um programa que o usuario digita um numero e o programa soma os numeros até o numero digitado
num = int(input("Digite um número: "))
N = 0
for i in range(1, num + 1):
  N += i
print(f"A soma dos números até {num} é: {N}")
```

```python
# 6 - Escrever um programa que o usuario digita uma palavra e o programa conta as letras da palavra
palavra = input("Digite uma palavra: ")
g = len(palavra)
print(f"A palavra '{palavra}' tem {g} letras.")
```

```python
# 7 - Escrever um programa que o usuario digita uma palavra e o programa imprimi a palavra invertida
palavra = input("Digite uma palavra: ")
palavra_i = palavra[::-1]
print(f"A palavra invertida é: {palavra_i}")
```

```python
# 8 - Escrever um programa que o usuario digita 4 numeros e o programa imprimi o maior
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))
num4 = int(input("Digite o quarto número: "))

maior = num1

if num2 > maior:
  maior = num2
if num3 > maior:
  maior = num3
if num4 > maior:
  maior = num4

print(f"O maior número digitado é: {maior}")
```

```python
# 9 - piramide numerica simetrica
# faça um programa que receba um numero N e imprima uma piramide perfeitamente centralizada
n = int(input("Digite um número para a pirâmide: "))

for i in range(1, n + 1):
    # Calculate leading spaces
    spaces = " " * (n - i)

    # Generate numbers for the ascending part
    nums_asc = "".join(str(j) for j in range(1, i + 1))

    # Generate numbers for the descending part (excluding the peak if i > 1)
    nums_desc = "".join(str(j) for j in range(i - 1, 0, -1))

    # Combine parts and print
    print(f"{spaces}{nums_asc}{nums_desc}")
```
